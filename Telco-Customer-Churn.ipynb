```python
X = df.drop('churn',axis=1)
y = df['churn']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder , StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
```

```python
numbers_1 = X.select_dtypes(include=['float64','int64']).columns
category_1 = X.select_dtypes(include=['object']).columns
```

```python
number_pipeline = Pipeline([('impute',SimpleImputer(strategy='median')),('scaler',StandardScaler())])
category_pipeline = Pipeline([('impute',SimpleImputer(strategy='most_frequent')),('encoder',OneHotEncoder(handle_unknown='ignore'))])
```

```python
transformer = ColumnTransformer([('numbers',number_pipeline,numbers_1),('category',category_pipeline,category_1)])
```

```python
transformer
```

```python
from imblearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from imblearn.over_sampling import SMOTE
```

```python
pipeline = Pipeline([
    
    ('transformer', transformer),
    
    ('smote', SMOTE(random_state=42)),
    
    ('model', LogisticRegression())
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score , mean_absolute_error
```

```python
accuracy_score(y_test,y_pred)
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
pipeline_rf = Pipeline([
    
    ('transformer', transformer),
    
    ('smote', SMOTE(random_state=42)),
    
    ('model', RandomForestClassifier(n_estimators=400,max_depth=20,
    min_samples_split=5, min_samples_leaf=2,
    random_state=42))
])
```

```python
pipeline_rf.fit(X_train,y_train)
```

```python
y_pred_rf = pipeline_rf.predict(X_test)
```

```python
accuracy_score(y_test,y_pred_rf)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test, y_pred_rf))
```
